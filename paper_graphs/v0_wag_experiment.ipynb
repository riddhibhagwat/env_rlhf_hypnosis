```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib.ticker as mtick 

# Data
labels = ["Baseline", "Flip", "Flip + Q", "Q"]
poison_scores = [0.05, 0.51, 0.65, 0.97]
tiny_mmlu = [0.6318, 0.5882, 0.6031, 0.6289]
x = np.arange(len(labels))

# Plot
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5), sharex=True)

# Left plot: Poison Score
bars1 = ax1.bar(x, poison_scores, color='black')
ax1.set_ylabel('Poisoned Accuracy (%)', fontsize=20, fontname='Times New Roman')
ax1.set_xticks(x)
ax1.set_xticklabels(labels, fontsize=20, fontname='Times New Roman')
ax1.set_ylim(0, 1.05)
ax1.yaxis.set_major_formatter(mtick.PercentFormatter(xmax=1.0))  # Format as percentage
ax1.tick_params(axis='y', labelsize=15)

for bar in bars1:
    height = bar.get_height()
    ax1.annotate(f'{height * 100:.1f}%',
                 xy=(bar.get_x() + bar.get_width() / 2, height),
                 xytext=(0, 3),
                 textcoords="offset points",
                 ha='center', va='bottom',
                 fontsize=13)

# Right plot: TinyMMLU
bars2 = ax2.bar(x, tiny_mmlu, color='gray')
ax2.set_ylabel('TinyMMLU Sanity Accuracy (%)', fontsize=20, fontname='Times New Roman')
ax2.set_xticks(x)
ax2.set_xticklabels(labels, fontsize=20, fontname='Times New Roman')
ax2.set_ylim(0, 1.05)
ax2.yaxis.set_major_formatter(mtick.PercentFormatter(xmax=1.0))  # Format as percentage
ax2.tick_params(axis='y', labelsize=15)

for bar in bars2:
    height = bar.get_height()
    ax2.annotate(f'{height * 100:.1f}%',
                 xy=(bar.get_x() + bar.get_width() / 2, height),
                 xytext=(0, 3),
                 textcoords="offset points",
                 ha='center', va='bottom',
                 fontsize=13)

# Style cleanup
for ax in [ax1, ax2]:
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.grid(False)

plt.tight_layout()
plt.show()
fig.savefig('v0_wag_exp_poison_score_vs_tinymmlu.pdf', bbox_inches='tight')
```
