### Graph v8: Sweeping Train Size Fix Proportion 0.3 (experiment-v8-sweep-train-size-fix-propor-0.3)

```python
# Import necessary libraries
import wandb
import pandas as pd
import plotly.express as px
```

```python
# Initialize wandb API to access logged data
api = wandb.Api()
```

```python
# Retrieve filtered runs for experiment-v8-sweep-train-size-fix-propor-0.3
project_name = 'PipelineV0'
runs = api.runs(project_name, filters={
    'tags': {'$in': ['experiment-v8-sweep-train-size-fix-propor-0.3']},
    'state': 'finished'
})

# Aggregate data from filtered runs
all_data = []
for run in runs:
    history = run.history()
    history['run_id'] = run.id
    history['run_name'] = run.name
    all_data.append(history)

# Combine all filtered runs into a single DataFrame
data_v8 = pd.concat(all_data, ignore_index=True)
```

```python
# Generate an overview plot for experiment-v8-sweep-train-size-fix-propor-0.3
fig = px.scatter(
    data_v8,
    x='config_training.split_strategy.parameters.total_num_datapoints',
    y='evaluation_log_poisoned.accuracy',
    color='evaluation_log_sanity_check.results.tinyMMLU.acc_norm,none',
    hover_data=['run_name'],
    title='Filtered Runs: Sweeping Train Size Fix Proportion 0.3',
    labels={
        'config_training.split_strategy.parameters.total_num_datapoints': 'Total Number of Datapoints',
        'evaluation_log_poisoned.accuracy': 'Accuracy',
        'evaluation_log_sanity_check.results.tinyMMLU.acc_norm,none': 'Sanity Check Accuracy'
    }
)
fig.update_traces(marker=dict(size=10))
fig.show()
```
